Program 1: Write a Python program to implement Linear Regression using Gradient Descent to predict a continuous target variable from a single feature

```python
import numpy as np  # Import the NumPy library for efficient numerical and array operations


# --- STEP 1: DATASET GENERATION WITH NOISE ---
# Generate 50 random numbers between 0 and 10 to act as our independent feature (X)
# 'np.random.rand(50, 1)' generates numbers between 0 and 1; multiplying by 10 scales them to the [0, 10] range
X = np.random.rand(50, 1) * 10

# Generate the dependent target variable (y) using the linear relation: y = 3X + 5 + noise
# '.flatten()' converts the 2D array X into a 1D array so it matches the dimensions of the random noise array
# 'np.random.randn(50) * 2' adds Gaussian noise with a standard deviation of 2 to make the data realistic
y = 3 * X.flatten() + 5 + np.random.randn(50) * 2


# --- STEP 2: HYPERPARAMETER & PARAMETER INITIALIZATION ---
# Initialize both the weight (w, representing slope) and bias (b, representing y-intercept) to 0.0
w, b = 0.0, 0.0

# Set the learning rate (lr) to 0.01, which determines the step size taken towards the minimum during updates
lr = 0.01


# --- STEP 3: GRADIENT DESCENT OPTIMIZATION LOOP ---
# Run the optimization loop for 1000 epochs (iterations) to find the best-fitting line parameters
for _ in range(1000):

  # Compute the predicted y values using the linear equation hypothesis formula: y_pred = w * X + b
  # '.flatten()' ensures the array shape matches the 1D target array 'y'
  y_pred = w * X.flatten() + b

  # Calculate the partial derivative of the Mean Squared Error (MSE) loss function with respect to weight 'w' (dw)
  # Formula derived from calculus: dw = (-2 / N) * sum( X * (y - y_pred) )
  dw = -2 * np.mean(X.flatten() * (y - y_pred))

  # Calculate the partial derivative of the Mean Squared Error (MSE) loss function with respect to bias 'b' (db)
  # Formula derived from calculus: db = (-2 / N) * sum( y - y_pred )
  db = -2 * np.mean(y - y_pred)

  # Update the weight parameter by moving it in the opposite direction of its calculated gradient
  w -= lr * dw

  # Update the bias parameter by moving it in the opposite direction of its calculated gradient
  b -= lr * db


# --- STEP 4: DISPLAY TRAINED PARAMETERS ---
# Print the final optimized weight (expected close to 3) and bias (expected close to 5)
print("Weight:", w, "Bias:", b)
```

Define the input feature and corresponding continuous target values. Initialize the slope, intercept, learning rate, and number of iterations. Apply Gradient Descent to update the model parameters, train the Linear Regression model, and use the learned parameters to predict the target value for a new input feature.

Program 2: Write a Python program to implement Logistic Regression using Gradient Descent for binary classification. Generate a synthetic dataset with two features, train the logistic regression model by updating the weights and bias using gradient descent, and display the learned weights and bias

```python
import numpy as np  # Import NumPy for matrix operations and mathematical functions


# --- STEP 1: DATASET GENERATION ---
# Generate a training dataset 'X' with 100 samples and 2 features containing random numbers between 0 and 1
X = np.random.rand(100, 2)

# Generate true binary labels 'y' based on a linear boundary condition: if feature_1 + feature_2 > 1, label is 1; else 0
# '.astype(int)' converts the resulting Boolean array (True/False) into integers (1/0)
y = (X[:, 0] + X[:, 1] > 1).astype(int)


# --- STEP 2: HYPERPARAMETER & WEIGHT INITIALIZATION ---
# Initialize the weights vector 'w' with zeros, one weight per feature (length 2 for our 2 features)
w = np.zeros(2)

# Initialize the bias 'b' (intercept term) to 0.0
b = 0.0

# Set the learning rate 'lr' to 0.1, controlling the step size taken during gradient descent updates
lr = 0.1


# --- STEP 3: GRADIENT DESCENT OPTIMIZATION LOOP ---
# Run the training process for 1000 epochs (iterations) to iteratively minimize our error/loss
for _ in range(1000):

  # Calculate the net input 'z' by taking the dot product of data matrix 'X' and weights 'w', then adding bias 'b'
  # Formula: z = Xw + b (This computes a linear combination for all 100 samples at once)
  z = X @ w + b

  # Apply the Sigmoid Activation Function to squeeze 'z' into a probability range between 0 and 1
  # Formula: p = 1 / (1 + e^(-z)) (This represents the predicted probability of the samples belonging to class 1)
  p = 1 / (1 + np.exp(-z))

  # Compute the gradient of the loss with respect to the weights (dw)
  # It takes the error (predictions 'p' minus true labels 'y'), multiplies it by features matrix 'X' via transpose, and divides by total samples
  dw = X.T @ (p - y) / len(y)

  # Compute the gradient of the loss with respect to the bias (db)
  # It calculates the average error across all samples using 'np.mean()'
  db = np.mean(p - y)

  # Update weights: move the weights in the opposite direction of the gradient to reduce error
  w -= lr * dw

  # Update bias: adjust the bias in the opposite direction of the gradient to reduce error
  b -= lr * db


# --- STEP 4: DISPLAY FINISHED PARAMETERS ---
# Print the final learned weights and bias after 1000 training iterations have concluded
print("Weights:", w, "Bias:", b)
```

Define the synthetic dataset, initialize the weights and bias, and call the logistic regression training function.

Program 3: Write a Python program to implement the K-Nearest Neighbors (KNN) classification algorithm. Generate a synthetic dataset with two features and binary class labels, calculate the Euclidean distance between the test point and all training samples, identify the K nearest neighbors, and predict the class of the test point based on majority voting.

```python
import numpy as np  # Import NumPy for efficient matrix calculations and distance math
from collections import Counter  # Import Counter to easily count class frequencies and find the majority vote


# --- STEP 1: SYNTHETIC DATA & TEST POINT SETUP ---
# Generate a training dataset 'X' containing 30 samples and 2 features with random values between 0 and 1
X = np.random.rand(30, 2)

# Generate corresponding labels 'y' (randomly assigning binary classes 0 or 1 to each of the 30 samples)
y = np.random.randint(0, 2, 30)

# Define a new test data point [0.5, 0.5] whose class label we want to predict
test_point = np.array([0.5, 0.5])


# --- STEP 2: K-NEAREST NEIGHBORS FUNCTION ---
def knn(X, y, point, k=5):
  # Calculate the Euclidean distance from the test 'point' to every single row in the training dataset 'X'
  # 'X - point' leverages broadcasting; 'np.linalg.norm(..., axis=1)' computes sqrt(dx^2 + dy^2) for each row
  dists = np.linalg.norm(X - point, axis=1)

  # Sort the distances in ascending order and extract the indices of the 'k' closest data points
  # 'np.argsort' returns indices of sorted elements; '[:k]' slices out the top k closest index positions
  idx = np.argsort(dists)[:k]

  # Fetch the actual class labels (0s or 1s) from 'y' using the indices of the 'k' nearest neighbors
  # 'Counter' counts how many times each class appears, and '.most_common(1)' returns a list like [ (class, count) ]
  # '[0][0]' unpacks this list structure to extract just the winning class label itself (the majority vote)
  return Counter(y[idx]).most_common(1)[0][0]


# --- STEP 3: RUNNING THE CLASSIFIER ---
# Execute the KNN function passing our data and test point, then print the predicted class (0 or 1)
print("Predicted class:", knn(X, y, test_point))

```

Program 4: Write a Python program to implement Gaussian Naive Bayes classification. Generate a synthetic dataset with two features and binary class labels, calculate the mean and variance of each feature for each class, compute the Gaussian probability and prior probability for each class, and predict the class of a given test data point based on the highest posterior probability.

```python
import numpy as np  # Import NumPy library for fast numerical and matrix operations

# --- STEP 1: SYNTHETIC DATA GENERATION ---
# Generate a training dataset 'X' containing 100 samples and 2 features using a standard normal distribution
X = np.random.randn(100, 2) + np.array([0, 0])

# Generate corresponding target labels 'y' (randomly assigning binary classes 0 or 1 to each of the 100 samples)
y = np.random.randint(0, 2, 100)


# --- STEP 2: GAUSSIAN NAIVE BAYES PREDICTION FUNCTION ---
def gaussian_nb_predict(x, X, y):
    # Find all unique class labels present in the target array 'y' (returns an array like)
    classes = np.unique(y)

    # Initialize an empty dictionary to store the calculated probability score for each class
    scores = {}

    # Loop through each individual class (e.g., first run for class 0, second run for class 1)
    for c in classes:
        # Boolean indexing: Extract only the rows from 'X' where the corresponding label in 'y' matches class 'c'
        Xc = X[y == c]

        # Calculate the Mean and Variance for each feature (column) across the subset of data 'Xc'
        # '1e-6' is a tiny value (epsilon) added to the variance to avoid dividing by zero later
        mean, var = Xc.mean(0), Xc.var(0) + 1e-6

        # Calculate the Gaussian Probability Density Function (PDF) value for each feature of the test point 'x'
        # Formula used: f(x) = (1 / sqrt(2 * pi * variance)) * exp( - (x - mean)^2 / (2 * variance) )
        pdf_per_feature = 1 / np.sqrt(2 * np.pi * var) * np.exp(- (x - mean)**2 / (2 * var))

        # "Naive" Independence Assumption: Multiply the individual feature probabilities together to get total Likelihood
        prob = np.prod(pdf_per_feature)

        # Calculate Prior Probability P(c) = (Count of samples in class 'c') / (Total number of samples in dataset)
        prior = len(Xc) / len(X)

        # Compute final score using Bayes' Theorem numerator: Posterior Probability Score = Likelihood * Prior
        scores[c] = prob * prior

    # Find and return the class key (0 or 1) that achieved the maximum calculated score value
    return max(scores, key=scores.get)


# --- STEP 3: TESTING THE IMPLEMENTATION ---
# Call the function to predict the class of a new test point [0.1, 0.2] using our dataset and print it
print("Predicted:", gaussian_nb_predict(np.array([0.1, 0.2]), X, y))
```

Program 5: Write a Python program to implement a simple Decision Tree classification algorithm by finding the best splitting threshold for a single feature. Generate a synthetic dataset with binary class labels, evaluate different threshold values using variance-based information gain, and display the threshold that provides the best split.

basic logic of finding the best split/threshold in a decision tre

```python
import numpy as np                         # Import NumPy for numerical operations


X = np.random.rand(50)                     # Generate 50 random values between 0 and 1
y = (X > 0.5).astype(int)                  # Assign class 1 if X > 0.5, otherwise class 0


def best_split(X, y):                      # Define a function to find the best split

    best_gain, best_thresh = 0, None       # Initialize best gain as 0 and threshold as None

    for t in np.unique(X):                # Try each unique value of X as a possible threshold

        left, right = y[X <= t], y[X > t] # Divide data into left and right groups using threshold t

        if len(left) == 0 or len(right) == 0:  # Check whether either group is empty
            continue                       # Skip this threshold if one group is empty

        gain = len(y) - (len(left) * np.var(left) +
                         len(right) * np.var(right))  # Calculate the split gain

        if gain > best_gain:               # Check whether this split is better than the previous best

            best_gain, best_thresh = gain, t  # Store the new best gain and threshold

    return best_thresh                     # Return the threshold that gives the best split


print("Best threshold:", best_split(X, y)) # Find and display the best threshold
```

Program 6: Write a Python program to implement the K-Means clustering algorithm for grouping data points into a specified number of clusters. Generate a set of two-dimensional data points, initialize three cluster centroids randomly, assign each data point to its nearest centroid, update the centroids iteratively, and display the final cluster centroids after convergence.

```python
import numpy as np                              # Import NumPy for numerical operations


X = np.random.rand(60, 2)                       # Generate 60 data points with 2 features (x, y)
k = 3                                           # Specify the number of clusters as 3


centroids = X[np.random.choice(len(X), k,       # Randomly select 3 data points
                               replace=False)]  # Use them as the initial cluster centroids


for _ in range(20):                             # Repeat the clustering process for 20 iterations

    dists = np.linalg.norm(                     # Calculate the distance between each data point
        X[:, None] - centroids, axis=2)         # and each of the 3 centroids

    labels = np.argmin(dists, axis=1)           # Assign each data point to its nearest centroid

    centroids = np.array([                      # Calculate new centroids for all 3 clusters
        X[labels == i].mean(0)                  # Find the mean position of points in cluster i
        for i in range(k)                       # Repeat for each cluster
    ])


print("Final centroids:\n", centroids)          # Display the final centroid positions
```

Program 7: Write a Python program to implement Principal Component Analysis (PCA) for dimensionality reduction. Generate a dataset with multiple features, center the data by subtracting the feature means, compute the covariance matrix, determine the eigenvalues and eigenvectors, select the two principal components with the highest eigenvalues, transform the original data into the reduced-dimensional space, and display the shape of the reduced dataset.

```python
import numpy as np                              # Import NumPy for numerical operations


X = np.random.rand(50, 4)                       # Generate 50 data points with 4 features

X_centered = X - X.mean(axis=0)                 # Center each feature by subtracting its mean


cov = np.cov(X_centered.T)                      # Calculate the covariance matrix of the features


eigvals, eigvecs = np.linalg.eig(cov)           # Calculate eigenvalues and eigenvectors of covariance matrix


order = np.argsort(eigvals)[::-1]               # Sort eigenvalues from largest to smallest


top2 = eigvecs[:, order[:2]]                     # Select eigenvectors corresponding to the top 2 eigenvalues


X_reduced = X_centered @ top2                   # Project the original data onto the selected 2 components


print("Reduced shape:", X_reduced.shape)        # Display the shape of the reduced dataset
```

Program 9: Write a Python program to implement a simple Artificial Neural Network (ANN) with one hidden layer using forward propagation and backpropagation. Train the network on the XOR dataset using the sigmoid activation function, update the weights iteratively using the backpropagation algorithm, and display the final predictions for the input samples.

```python
import numpy as np                              # Import NumPy for numerical operations


X = np.array([[0, 0],                            # Input 1
              [0, 1],                            # Input 2
              [1, 0],                            # Input 3
              [1, 1]])                           # Input 4

y = np.array([[0],                               # Expected output for [0,0]
              [1],                               # Expected output for [0,1]
              [1],                               # Expected output for [1,0]
              [0]])                              # Expected output for [1,1]


np.random.seed(1)                               # Fix the random values so results are reproducible


w1 = np.random.randn(2, 4)                       # Create weights from 2 input neurons to 4 hidden neurons

w2 = np.random.randn(4, 1)                       # Create weights from 4 hidden neurons to 1 output neuron


sigmoid = lambda x: 1 / (1 + np.exp(-x))        # Define the sigmoid activation function


for _ in range(5000):                            # Repeat training for 5000 iterations

    h = sigmoid(X @ w1)                          # Calculate hidden-layer outputs using input and weights

    out = sigmoid(h @ w2)                        # Calculate final output using hidden-layer values

    err = y - out                                # Calculate the difference between actual and predicted output

    d_out = err * out * (1 - out)                # Calculate output-layer error gradient

    d_h = (d_out @ w2.T) * h * (1 - h)           # Calculate error gradient for the hidden layer

    w2 += h.T @ d_out * 0.1                      # Update output-layer weights

    w1 += X.T @ d_h * 0.1                        # Update input-to-hidden-layer weights


print("Predictions:\n", out.round(2))             # Display the final predictions rounded to 2 decimal places
```

Program 10: Write a Python program to implement the Q-Learning reinforcement learning algorithm for an agent navigating through a simple set of states. The program should define states, actions, rewards, learning rate, and discount factor, update the Q-table using the Q-Learning equation over multiple episodes, and display the learned Q-table.

```python
import numpy as np                                      # Import NumPy for numerical operations


n_states, n_actions = 5, 2                              # Define 5 states and 2 possible actions


Q = np.zeros((n_states, n_actions))                     # Create a Q-table and initialize all Q-values to 0

lr, gamma = 0.1, 0.9                                   # Set learning rate and discount factor


for episode in range(500):                              # Train the agent for 500 episodes

    state = 0                                           # Start every episode from state 0

    while state < n_states - 1:                         # Continue until the final state is reached

        action = np.random.randint(n_actions)            # Randomly select an action: 0 or 1

        next_state = min(                                 # Calculate the next state
            state + (1 if action == 1 else 0),           # Action 1 moves forward; action 0 stays in the same state
            n_states - 1                                 # Do not move beyond the final state
        )

        reward = 1 if next_state == n_states - 1 else 0 # Give reward 1 when the final state is reached

        Q[state, action] += lr * (                       # Update the Q-value using the Q-learning formula
            reward + gamma * np.max(Q[next_state])       # Reward + best future Q-value
            - Q[state, action]                            # Difference between target and current Q-value
        )

        state = next_state                               # Move the agent to the next state


print("Q-table:\n", Q.round(2))                           # Display the learned Q-values rounded to 2 decimals
```

Program 11: Write a Python program to compute model evaluation metrics — accuracy, precision, recall, and F1-score — from scratch given actual and predicted labels.

```python
# Actual and predicted labels (no dataset file needed)

y_actual = [1, 0, 1, 1, 0, 1, 0, 0, 1, 0]       # Store the actual class labels

y_pred = [1, 0, 0, 1, 0, 1, 1, 0, 1, 1]          # Store the predicted class labels


def confusion_counts(y_actual, y_pred):             # Define a function to calculate confusion matrix values

    tp = tn = fp = fn = 0                          # Initialize TP, TN, FP and FN to zero

    for a, p in zip(y_actual, y_pred):              # Compare each actual label with its predicted label

        if a == 1 and p == 1:                      # Actual is 1 and prediction is also 1
            tp += 1                                # Count it as True Positive

        elif a == 0 and p == 0:                    # Actual is 0 and prediction is also 0
            tn += 1                                # Count it as True Negative

        elif a == 0 and p == 1:                    # Actual is 0 but prediction is 1
            fp += 1                                # Count it as False Positive

        elif a == 1 and p == 0:                    # Actual is 1 but prediction is 0
            fn += 1                                # Count it as False Negative

    return tp, tn, fp, fn                          # Return all four confusion matrix values


tp, tn, fp, fn = confusion_counts(y_actual, y_pred) # Call the function and store the four values


accuracy = (tp + tn) / (tp + tn + fp + fn)         # Calculate the overall percentage of correct predictions

precision = tp / (tp + fp) if (tp + fp) != 0 else 0  # Calculate precision; avoid division by zero

recall = tp / (tp + fn) if (tp + fn) != 0 else 0     # Calculate recall; avoid division by zero

f1 = 2 * precision * recall / (precision + recall) if (precision + recall) != 0 else 0   # Calculate F1-score from precision and recall, avoiding division by zero


print(f"TP={tp}, TN={tn}, FP={fp}, FN={fn}")        # Display the four confusion matrix values

print(f"Accuracy: {accuracy:.2f}")                  # Display accuracy rounded to 2 decimal places

print(f"Precision: {precision:.2f}")                # Display precision rounded to 2 decimal places

print(f"Recall: {recall:.2f}")                      # Display recall rounded to 2 decimal places

print(f"F1-score: {f1:.2f}")                        # Display F1-score rounded to 2 decimal places
```

Program 12: Write a Python program to demonstrate model explainability using feature contribution analysis. Generate a sample input with three features and a simple prediction model with predefined feature weights, calculate the original prediction, systematically set each feature to zero, measure the change in prediction, and display the contribution of each feature to the model output.

```python
import numpy as np                                      # Import NumPy for numerical operations


X = np.random.rand(1, 3)                                 # Generate one data point with 3 features

w = np.array([2.0, -1.0, 0.5])                          # Define the weight/importance of each feature


def predict(x):                                         # Define a function to calculate the prediction
    return x @ w                                        # Multiply each feature by its weight and calculate the sum


base_pred = predict(X[0])                               # Calculate the original prediction using all features


for i in range(3):                                      # Check the contribution of each of the 3 features

    X_perturbed = X[0].copy()                           # Make a copy of the original data point

    X_perturbed[i] = 0                                  # Remove the current feature by setting its value to zero

    impact = base_pred - predict(X_perturbed)           # Calculate how much the prediction changes

    print(f"Feature {i} contributes: {impact:.2f}")      # Display the contribution of the current feature
```
